#### Test 1(통과): $\qquad \hat U \hat H (\mathbf{k}) \hat U ^\dagger = \hat H (\hat R \mathbf{k}), \qquad \vec v ( \hat R \mathbf k) = \hat{U}\vec v (\mathbf k)$

```julia
using Random

include("χ0.jl")

mu = 1.0
Ham_sym = C4v_RashbaHubbard_sym()
ops = Ham_sym.ops
R = Ham_sym.R
U = Ham_sym.U
k = rand(2)

# "프로젝터 거리를 이용한 위상-불변 테스트"
projector(v) = v * v'                        # v assumed normalized
projector_distance(v1, v2) = opnorm(projector(v1) - projector(v2), 2)
# projector_distance(vR, U*vk) ≈ 0 이면 동일 상태로 판단


tol = 1e-12
for op_idx in eachindex(ops)
    println(ops[op_idx])
    H_k = C4v_RashbaHubbard_Ham(k, mu)
    Rk = R[op_idx] * k
    H_Rk = C4v_RashbaHubbard_Ham(Rk, mu)
    u = U[op_idx]
    # identity 1 비교
    sym_H_k = u * H_k * u'
    println("Hamiltonian diff. (isapprox(atol, rtol)): ", isapprox(H_Rk,sym_H_k; atol = tol, rtol = tol))
    # identity 2 비교
    _, eigen_k = eigen(Hermitian(H_k))
    eigen_k1 = eigen_k[:, 1]
    p1 = projector(u * eigen_k1)
    _, eigen_Rk = eigen(Hermitian(H_Rk))
    eigen_Rk1 = eigen_Rk[:,1]
    p2 = projector(eigen_Rk1)
    println("Eigenvector diff. (isapprox(atol, rtol)): ", isapprox(p1, p2; atol = tol, rtol = tol))
end
```
